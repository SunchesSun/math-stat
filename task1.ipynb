# Эмпирическая функция распределения и выборочная квантиль

## Задание

Необходимо составить максимально лаконичный код, который строит выборочную функцию распределения для заданной выборки и оценивает выборочную квантиль для заданной вероятности.

## Решение

Размер выборки `n`, среднее `mu`, стандартное отклонение `sigma` и вероятность квантили `p`. Массив `x` - выборка.

```python
import numpy as np, matplotlib.pyplot as plt
from scipy.stats import norm

n = 50
mu = 0 # среднее 
sigma = 1 # стандартное отколнение
p = 0.5
x = np.random.normal(mu, sigma, n)

t = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
plt.ecdf(x, label='выборочная $F_n(x)$')
plt.plot(t, norm.cdf(t, mu, sigma), label='теоретическая $F(x)$')
plt.legend()
plt.show()

print(np.quantile(x, p, method='inverted_cdf'))
```
